# Data Preprocessing

### -> Importing the necessary libraries

```python
import numpy as np # type: ignore
import pandas as pd # type: ignore
import matplotlib.pyplot as plt # type: ignore
```

### -> Loading and reading the dataset

```python
df = pd.read_csv('Data.csv')
```

```python
df
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

### -> Handling missing data - using mean imputation (with Sklearn)

```python
from sklearn.impute import SimpleImputer # type: ignore

imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
# Replacing missing values in the selected columns with the mean
X[:, 1:3] = imputer.fit_transform(X[:, 1:3])
```

```python
X
```

### -> Encoding categorical data (with Sklearn) using two classes:
- ColumnTransformer() class
- OneHotEncoder() class

```python
from sklearn.compose import ColumnTransformer # type: ignore
from sklearn.preprocessing import OneHotEncoder # type: ignore

# Creating a ColumnTransformer object
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [0])], remainder='passthrough', verbose=True)
X = np.array(ct.fit_transform(X))
```

```python
X
```

### -> Encoding the dependent variable (with Sklearn) using the same two class:
- ColumnTransformer() class
- OneHotEncoder() class

```python
from sklearn.compose import ColumnTransformer # type: ignore
from sklearn.preprocessing import OneHotEncoder # type: ignore

ct2 = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])], remainder='passthrough', verbose=True)
y = np.array(ct2.fit_transform(X))
```

```python
y
```

### -> Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split # type: ignore

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

### -> Feature Scaling using StandardScaler (with Sklearn) - Standardization Approach

```python
from sklearn.preprocessing import StandardScaler # type: ignore

sc = StandardScaler()

X_train[:, 3:] = sc.fit_transform(X_train[:, 3:])
X_test[:, 3:] = sc.transform(X_test[:, 3:])
```
