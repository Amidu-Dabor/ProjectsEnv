## NetflixDataCleaner Class Definition

```python
# NetflixDataCleaner Class Definition

class NetflixDataCleaner:
    def __init__(self, file_path):
        """Initialize with dataset loading"""
        self.file_path = file_path
        self.df = None
        self.original_shape = None
        self.load_data()
        
    def load_data(self):
        """Load the Netflix dataset from CSV"""

        self.df = pd.read_csv(self.file_path)
        self.original_shape = self.df.shape
        print(f"Dataset loaded: {self.original_shape[0]} rows, {self.original_shape[1]} columns")
    
    def explore_structure(self):
        """Explore basic dataset structure"""
        print("="*50)
        print("DATASET STRUCTURE EXPLORATION")
        print("="*50)
        
        # Basic info
        print(f"Shape (rows x columns): {self.df.shape}")
        print(f"Memory usage: {self.df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
        
        # Column names
        print("\nColumns in dataset:")
        for i, col in enumerate(self.df.columns, 1):
            print(f"  {i}. {col}")
        
        # Data types
        print("\nData types:")
        print(self.df.dtypes)
        
        # First few rows
        print("\nFirst 5 rows:")
        print(self.df.head())
        
        # Dataset info
        print("\nDataset info:")
        self.df.info()

    def discover_data_issues(self):
        """Discover data quality issues"""
        print("="*50)
        print("DATA DISCOVERY")
        print("="*50)
        
        # Missing values analysis
        print("Missing values per column:")
        missing_values = self.df.isnull().sum()
        for col, count in missing_values.items():
            if count > 0:
                percentage = (count / len(self.df)) * 100
                print(f"  {col}: {count} ({percentage:.1f}%)")
        
        print(f"\nTotal missing values: {missing_values.sum()}")
        
        # Duplicate analysis
        duplicates = self.df.duplicated().sum()
        print(f"Duplicate rows: {duplicates}")
        
        # Data type issues
        print("\nData type analysis:")
        print(self.df.dtypes.value_counts())
        
        # Check for inconsistencies
        print("\nData quality checks:")
        
        # Check date consistency
        temp_dates = pd.to_datetime(self.df['date_added'], errors='coerce')
        date_issues = (temp_dates.dt.year < self.df['release_year']).sum()
        print(f"  Date inconsistencies (added before release): {date_issues}")
        
        # Content type distribution
        print(f"\nContent type distribution:")
        print(self.df['type'].value_counts())
        
        # Statistical summary for numeric columns
        print(f"\nNumeric columns summary:")
        print(self.df.describe())

print("NetflixDataCleaner class defined successfully!")
```

## Cleaning methods to add t the NetflixDataCleaner class

```python
# Adding cleaning methods to the NetflixDataCleaner class
def clean_data(self):
    """Clean the dataset step by step"""
    print("="*50)
    print("DATA CLEANING")
    print("="*50)
    
    # Remove duplicates
    print("Removing duplicates...")
    initial_rows = len(self.df)
    self.df = self.df.drop_duplicates()
    removed = initial_rows - len(self.df)
    print(f"   Removed {removed} duplicate rows")
    
    # Drop description column (not needed for analysis)
    print("Removing unnecessary columns...")
    self.df = self.df.drop(columns=['description'])
    print("   Dropped 'description' column")
    
    # Fix date format
    print("Converting date format...")
    self.df['date_added'] = pd.to_datetime(self.df['date_added'], format='mixed')
    print("   Converted date_added to datetime")
    
    # Parse duration column
    print("Parsing duration column...")
    duration_parts = self.df['duration'].str.extract(r'(\d+)\s*(\w+)')
    self.df['duration_value'] = pd.to_numeric(duration_parts[0])
    self.df['duration_unit'] = duration_parts[1]
    print("   Created duration_value and duration_unit columns")
    
    # Handle missing directors using cast relationships
    print("Handling missing directors...")
    self._fix_missing_directors()
    
    # Handle missing countries using director relationships
    print("Handling missing countries...")
    self._fix_missing_countries()
    
    # Handle other missing values
    print("Handling remaining missing values...")
    self.df['cast'] = self.df['cast'].fillna('Not Given')
    
    # Drop rows with missing critical data
    before_drop = len(self.df)
    self.df = self.df.dropna(subset=['date_added', 'rating', 'duration'])
    after_drop = len(self.df)
    print(f"   Dropped {before_drop - after_drop} rows with missing critical data")
    
    print("Data cleaning completed!")

def _fix_missing_directors(self):
    """Fix missing director values using cast relationships"""
    # Create director-cast pairs for frequent collaborations
    valid_data = self.df[self.df['director'].notna() & self.df['cast'].notna()]
    dir_cast_pairs = valid_data['director'] + '---' + valid_data['cast']
    
    # Find pairs that appear 3+ times
    frequent_pairs = dir_cast_pairs.value_counts()
    frequent_pairs = frequent_pairs[frequent_pairs >= 3]
    
    # Create mapping dictionary
    director_cast_map = {}
    for pair in frequent_pairs.index:
        director, cast = pair.split('---')
        director_cast_map[cast] = director
    
    # Fill missing directors
    missing_directors = self.df['director'].isna()
    for cast, director in director_cast_map.items():
        mask = missing_directors & (self.df['cast'] == cast)
        self.df.loc[mask, 'director'] = director
    
    # Fill remaining with 'Not Given'
    self.df['director'] = self.df['director'].fillna('Not Given')
    print(f"   Fixed {len(director_cast_map)} director values using cast relationships")

def _fix_missing_countries(self):
    """Fix missing country values using director relationships"""
    # Create director-country mapping
    valid_data = self.df[self.df['director'].notna() & self.df['country'].notna()]
    director_country_map = dict(zip(valid_data['director'], valid_data['country']))
    
    # Fill missing countries
    missing_countries = self.df['country'].isna()
    for director, country in director_country_map.items():
        mask = missing_countries & (self.df['director'] == director)
        self.df.loc[mask, 'country'] = country
    
    # Fill remaining with 'Not Given'
    self.df['country'] = self.df['country'].fillna('Not Given')
    print(f"   Fixed {len(director_country_map)} country values using director relationships")

# Attach methods to class
NetflixDataCleaner.clean_data = clean_data
NetflixDataCleaner._fix_missing_directors = _fix_missing_directors
NetflixDataCleaner._fix_missing_countries = _fix_missing_countries

print("Data cleaning methods added to NetflixDataCleaner class!")
```

## Transform and analysis methods

```python
# Transform and analysis methods, including feature extraction using statsmodels api

def transform_and_enrich(self):
    """Transform and enrich the dataset"""
    print("="*50)
    print("DATA TRANSFORMATION AND ENRICHMENT")
    print("="*50)
    
    # Basic feature extraction
    print("Creating basic derived features...")
    
    # Date features
    self.df['year_added'] = self.df['date_added'].dt.year
    self.df['month_added'] = self.df['date_added'].dt.month
    self.df['day_of_week'] = self.df['date_added'].dt.day_name()
    
    # Content age
    current_year = dt.datetime.now().year
    self.df['content_age'] = current_year - self.df['release_year']
    
    # Time to Netflix
    self.df['years_to_netflix'] = self.df['year_added'] - self.df['release_year']
    
    # Genre count
    self.df['genre_count'] = self.df['listed_in'].str.count(',') + 1
    
    # Cast count
    self.df['cast_count'] = self.df['cast'].apply(
        lambda x: 0 if x == 'Not Given' else x.count(',') + 1
    )
    
    print("   Created: year_added, month_added, day_of_week, content_age")
    print("   Created: years_to_netflix, genre_count, cast_count")
    
    # Data Filtering
    print("\nData filtering...")
    recent_content = self.df[self.df['release_year'] >= 2020]
    print(f"   Recent content (2020+): {len(recent_content)} records")
    
    movies = self.df[self.df['type'] == 'Movie']
    tv_shows = self.df[self.df['type'] == 'TV Show']
    print(f"   Movies: {len(movies)} | TV Shows: {len(tv_shows)}")
    
    # Data Sorting
    print("\nData sorting...")
    newest_first = self.df.sort_values('release_year', ascending=False)
    print(f"   Sorted by release year (newest first)")

    self.df['title_length'] = self.df['title'].str.len()
    longest_titles = self.df.sort_values('title_length', ascending=False)
    print(f"   Sorted by title length")
    
    # Data Grouping
    print("\nData grouping...")
    
    # Group by content type
    type_summary = self.df.groupby('type').agg({
        'title': 'count',
        'release_year': 'mean',
        'duration_value': 'mean'
    }).round(2)
    print("   Content type summary:")
    print(type_summary)
    
    # Group by year added
    yearly_additions = self.df.groupby('year_added')['title'].count()
    print(f"\n   Content additions by year:")
    print(yearly_additions.tail())
    
    # Top countries
    top_countries = self.df['country'].value_counts().head()
    print(f"\n   Top 5 countries:")
    print(top_countries)

def extract_predictive_features(self):
    """Extract and rank features based on their predictive power using OLS regression"""
    print("\n5. PREDICTIVE FEATURE EXTRACTION USING REGRESSION")
    print("-" * 50)
    
    print("Extracting features that predict content duration using statistical analysis...")
    
    # Prepare features for extraction analysis
    feature_cols = ['release_year', 'content_age', 'genre_count', 'cast_count', 'years_to_netflix']
    
    # Focus on movies for consistent analysis
    movies_df = self.df[self.df['type'] == 'Movie'].copy()
    
    # Extract clean feature set
    X = movies_df[feature_cols].dropna()
    y = movies_df.loc[X.index, 'duration_value']
    
    print(f"   Analyzing {len(feature_cols)} extracted features across {len(X)} movies")
    print(f"   Target variable: duration_value (movie length in minutes)")
    
    # Add constant for regression analysis
    X_with_const = sm.add_constant(X)
    
    # Perform feature extraction analysis
    print(f"\n   Fitting OLS regression model for feature extraction...")
    model = sm.OLS(y, X_with_const).fit()
    
    # Display extraction results
    print("\nSTATISTICAL FEATURE EXTRACTION RESULTS:")
    print("="*60)
    print(model.summary())
    
    # Extract key insights about features
    print(f"\nEXTRACTED FEATURE INSIGHTS:")
    print("-" * 30)
    print(f"Feature set explains {model.rsquared:.1%} of duration patterns")
    print(f"Model statistical significance: F = {model.fvalue:.2f}")
    print(f"Analyzed {len(feature_cols)} candidate features")
    
    print(f"\nMOST PREDICTIVE EXTRACTED FEATURES (p < 0.05):")
    print("-" * 50)
    significant_features = []
    for feature in model.pvalues.index:
        if model.pvalues[feature] < 0.05 and feature != 'const':
            coef = model.params[feature]
            p_val = model.pvalues[feature]
            significant_features.append(feature)
            direction = "increases" if coef > 0 else "decreases"
            print(f"{feature}: {direction} duration by {abs(coef):.2f} min (p = {p_val:.4f})")
    
    if not significant_features:
        print("  No features reached statistical significance (p < 0.05)")
    
    print(f"\nFEATURE EXTRACTION SUMMARY:")
    print("-" * 30)
    print(f"Total features analyzed: {len(feature_cols)}")
    print(f"Statistically significant features: {len(significant_features)}")
    print(f"Model predictive power: {model.rsquared:.1%}")
    
    return significant_features, model

# Attach methods to class
NetflixDataCleaner.transform_and_enrich = transform_and_enrich
NetflixDataCleaner.extract_predictive_features = extract_predictive_features

print("Transform and feature extraction methods added to NetflixDataCleaner class!")
```

## Validation and export methods

```python
# Validation and export methods

def validate_dataset(self):
    """
    Validate the cleaned and transformed dataset.

    This function performs several validation checks on the dataset, including data type validation,
    completeness checks, and business logic validation. It also provides a summary of the final dataset
    statistics and a sample of the data.

    Parameters:
    self : object
        The instance of the class containing the dataset to be validated.

    Returns:
    None
    """
    print("="*50)
    print("DATA VALIDATION")
    print("="*50)
    
    # Check data types
    print("Data type validation...")
    expected_types = {
        'date_added': 'datetime64[ns]',
        'duration_value': ['int64', 'float64'],
        'year_added': ['int64', 'float64']
    }
    
    for col, expected in expected_types.items():
        actual = str(self.df[col].dtype)
        if isinstance(expected, list):
            is_valid = actual in expected
        else:
            is_valid = actual == expected
        status = "✓" if is_valid else "✗"
        print(f"   {status} {col}: {actual}")
    
    # Check completeness
    print("\nData completeness check...")
    missing_values = self.df.isnull().sum()
    if missing_values.sum() == 0:
        print("   No missing values found")
    else:
        print("   Missing values found:")
        for col, count in missing_values[missing_values > 0].items():
            print(f"     {col}: {count}")
    
    # Business logic validation
    print("\nBusiness logic validation...")
    
    # Check for reasonable release years
    current_year = dt.datetime.now().year
    old_content = (self.df['release_year'] < 1900).sum()
    future_content = (self.df['release_year'] > current_year).sum()
    print(f"   Content before 1900: {old_content}")
    print(f"   Future content: {future_content}")
    
    # Check duration reasonableness
    movie_duration = self.df[self.df['type'] == 'Movie']['duration_value']
    unreasonable_movies = ((movie_duration < 30) | (movie_duration > 300)).sum()
    print(f"   Movies with unreasonable duration: {unreasonable_movies}")
    
    # Check date consistency
    date_issues = (self.df['date_added'].dt.year < self.df['release_year']).sum()
    print(f"   Date inconsistencies: {date_issues}")
    
    # Sample final data
    print("\nSample of final dataset:")
    sample_cols = ['title', 'type', 'release_year', 'year_added', 'genre_count']
    print(self.df[sample_cols].sample(5).to_string(index=False))
    
    # Final statistics
    print(f"\nFinal dataset statistics:")
    print(f"   Rows: {len(self.df)}")
    print(f"   Columns: {len(self.df.columns)}")
    print(f"   Memory usage: {self.df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
    reduction = ((self.original_shape[0] - len(self.df)) / self.original_shape[0]) * 100
    print(f"   Data reduction: {reduction:.1f}%")

def export_cleaned_data(self, output_path='cleaned_netflix.csv'):
    """Export the cleaned dataset"""
    print("="*50)
    print("DATA EXPORT")
    print("="*50)
    
    # Reset index for clean export
    self.df = self.df.reset_index(drop=True)
    
    # Export to CSV
    self.df.to_csv(output_path, index=False)
    
    print(f"Dataset exported to: {output_path}")
    print(f"Records exported: {len(self.df)}")
    print(f"Columns exported: {len(self.df.columns)}")
    
    # Show final column list
    print("\nFinal columns in exported dataset:")
    for i, col in enumerate(self.df.columns, 1):
        print(f"  {i:2d}. {col}")
    
    print("\nData wrangling process completed successfully!")

# Attach methods to class
NetflixDataCleaner.validate_dataset = validate_dataset
NetflixDataCleaner.export_cleaned_data = export_cleaned_data
```
