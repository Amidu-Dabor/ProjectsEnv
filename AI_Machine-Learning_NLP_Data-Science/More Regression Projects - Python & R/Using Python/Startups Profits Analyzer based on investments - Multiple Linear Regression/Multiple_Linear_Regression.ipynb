# Multiple Linear Regression

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
sns.set_theme()
```

## Loading the dataset

```python
df = pd.read_csv('50_Startups.csv')
df.head()
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values # The target column ('Profit')
```

## Encoding categorical data

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
X
```

## Splitting the data into training and test sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)


# There is no need for 'feature scaling' in 'Multiple Linear Regression'
# This is because the biases of the inputs (bias of each input) compensate for 'feature scaling'
# So, we skip the 'Feature Scaling' step.
```

## Training the Multiple Linear Regression

```python
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)


# The 'Dummy Variable' Trap (Automatically handled in 'Multiple Linear Regression'):
# The 'Multiple Regression' model (class) automatically avoids the 'Dummy Variable' trap.
```

## Predicting the test set results -> Evaluating the model

```python
# We cannot plot/create a graph for a 'Multiple Linear Regression' model as in 'Simple Linear Regression',
# because it does not produce a clear 'linear' or 'non-linear' relationship, which is not possible to visualize as humans.
# We will, instead, display '2 vectors', where one vector represents the 'Profit' (dependent variable) and 
# the other vector represents the 'Investments' (independent variables).

y_pred = regressor.predict(X_test)
np.set_printoptions(precision=2) # Precision of the output to 2 decimal places
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

## Evaluating the model - R-squared, Adjusted R-squared
* The 'Adjusted R-squared' value is more reliable than the 'R-squared' value,
* as it takes into account the number of independent variables.

#### R-squared Formula:
- $R^2 = 1 -\frac{SS_{res}}{SS_{total}}$
* $SS_{res}$ # = Sum of Squares of Residuals -> $SUM(y_{i} - \hat{y_{i}})^2$
* $SS_{total}$ # = Sum of Squares of Totals -> $SUM(y_{i} - \bar{y_{i}})^2$
* $\hat{y_{i}}$ = The predicted values of y.
* $\bar{y_{i}}$ = The mean of y.

#### Adjusted R-squared Formula:
- $R^2_{adj.} = 1 - (1 - R^2)*\frac{n - 1}{n - k - 1}$
* n = len(y) # Number of observations
* k = X.shape[1] # Number of features (independent variables)

## Calculating the R-squared and Adjusted R-squared

```python
# Calculating the R-squared
# r2 = regressor.score(y_test, y_pred)
# # Calculating the Adjusted R-squared
# n = X.shape[0]
# k = X.shape[1]
# adjusted_r2 = 1-(1-r2)*(n-1)/(n-k-1)
# np.set_printoptions(precision=2)

from sklearn.metrics import r2_score

r2 = r2_score(y_true=y_test, y_pred=y_pred)
n = X.shape[0]
k = X.shape[1]
adjusted_r2 = 1-(1-r2)*(n-1)/(n-k-1)
print(f'R-squared: {r2:.3f}, Adjusted R-squared: {adjusted_r2:.3f}')
```
