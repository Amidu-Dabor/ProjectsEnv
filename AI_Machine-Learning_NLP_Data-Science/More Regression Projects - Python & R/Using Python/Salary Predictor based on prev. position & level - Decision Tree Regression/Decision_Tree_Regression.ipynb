# Decision Tree Regression
- To Predict the previous salary of a new Employee (recruit) and compare to the their expected salary
- Decision Trees are a type of supervised learning algorithm used for both regression and classification tasks.
- Decision Trees work by recursively splitting the data into smaller subsets based on the most important features.
- Decision Trees are easy to understand and interpret, making them popular for data exploration and analysis.
- Decision Trees can handle both numerical and categorical data, making them suitable for a wide range of applications.
- Decision Trees are prone to overfitting if the tree is too deep or if the training data is noisy.
- Decision Trees are not suitable for datasets with a small number of samples or high-dimensional data.

#### Key Characteristics:
1. Non-Linear Model: Captures non-linear relationships by creating partitions.
2. Recursive Splitting: Splits data based on the best feature and threshold at each step.
3. Interpretability: Easy to understand and interpret as it mimics human decision-making.

#### Steps in Decision Tree Regression
1. The algorithm evaluates every feature and its possible split points to identify the split that minimizes the loss function (e.g., variance reduction).
2. Splits are created recursively until a stopping condition is met (e.g., maximum depth or minimum samples per leaf).
3. Predictions for a new input are made by following the decision rules through the tree until reaching a leaf node.

#### Advantages
1. Non-Parametric: Makes no assumptions about data distribution.
2. Handles Non-Linearity: Can model complex relationships without requiring transformation of data.
3. Interpretability: Easy to visualize and explain the decision-making process.
4. Handles Mixed Data Types: Works well with both numerical and categorical data.

#### Limitations
1. Overfitting: Prone to overfitting, especially with deep trees or small datasets.
2. Instability: Small changes in data can lead to significant changes in the tree structure.
3. Biased Splits: Splits can favor features with many levels (e.g., categorical features with high cardinality).

#### When to Use
1. When interpretability is important.
2. When the relationship between features and target is non-linear or complex.
3. When you want to analyze feature importance or identify patterns in data subsets.
* For larger datasets or to avoid overfitting, ensemble methods like Random Forests or Gradient Boosting can be used as an extension of Decision Trees.

## Importing the libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
```

## Loading the dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
dataset
```

```python
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

## Training the Decision Tree Regression model on the whole dataset

```python
from sklearn.tree import DecisionTreeRegressor

regressor = DecisionTreeRegressor(random_state = 0)
regressor.fit(X, y)
```

## Predicting a new result

```python
regressor.predict([[6.5]])
```

## Visualising the Decision Tree Regression results (higher resolution)

```python
plt.scatter(X, y, color ='red')
X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.plot(X_grid, regressor.predict(X_grid), color = 'blue')
plt.title('Truth or Bluff (Decision Tree Regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```
