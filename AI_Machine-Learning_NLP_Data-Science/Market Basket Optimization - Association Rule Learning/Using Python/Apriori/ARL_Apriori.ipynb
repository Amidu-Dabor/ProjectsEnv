# Association Rule Learning (Apriori Algorithm) - ARL
- Association rule learning involves identifying strong associations between different items in a dataset.

- A rule is defined as an implication: if item A is purchased, then item B is likely to be purchased.

#### Apriori Algorithm Steps:
1. Data Preparation:
    - Collect a dataset of transactions, where each transaction is a set of items.

    - Convert the dataset into a suitable format, such as a binary matrix or a transaction database.

2. Generate Frequent Itemsets:
    - Identify the most frequent itemsets in the dataset using a suitable algorithm, such as Apriori.

    - Frequent itemsets are those that appear in at least a certain number of transactions.

#### Types of Association Rule Learning Algorithms:
1. Apriori:
    - Iteratively identifies frequent itemsets using a bottom-up approach.

    - Filters combinations based on minimum support and generates rules with confidence thresholds.

    - Example: In a grocery store, Apriori identifies that people buying bread often buy butter.

2. Eclat (Equivalence Class Clustering and Bottom-up Lattice Traversal):
    - Uses a depth-first search approach with vertical dataset representation.

    - Faster than Apriori for large datasets.

    - Example: Eclat quickly identifies frequent patterns like {milk, cereal} in sales data.

3. FP-Growth (Frequent Pattern Growth):
    - Constructs a compact tree structure (FP-tree) for storing itemsets.

    - Avoids candidate generation like Apriori.
    
    - Example: FP-Growth detects patterns like {diapers, beer} in sales efficiently.

## Importing the necessary libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()
```

## Loading the dataset

```python
dataset = pd.read_csv('Market_Basket_Optimisation.csv', header = None) # The 'header = None' indicates no column names.
dataset
```

## Data Preprocessing

```python
transactions = []
for i in range(0, dataset.shape[0]):
  temp = [str(dataset.values[i,j]) for j in range(0, dataset.shape[1])] # 'i = row index', 'j = column index'
  temp = sorted(list(set(temp)))
  transactions.append(temp)
```

```python
transactions
```

## Training the Apriori model on the dataset

```python
from apyori import apriori

# Training the Apriori model on the dataset
rules = apriori(transactions = transactions, 
                min_support = 0.003, 
                min_confidence = 0.2, 
                min_lift = 3, 
                min_length = 2, 
                max_length = 2)
```

## Visualising the Results

## Dsiplaying the first results 

```python
# Displaying the rules in a readable format
# Putting the results well organised into a Pandas DataFrame
results = list(rules)
results
```

## Putting the results well organised into a Pandas DataFrame

```python
def inspect(results):
    left_hand_side_of_rule  = [tuple(result[2][0][0])[0] for result in results]
    right_hand_side_of_rule = [tuple(result[2][0][1])[0] for result in results]
    supports                = [result[1] for result in results]
    confidences             = [result[2][0][2] for result in results]
    lifts                   = [result[2][0][3] for result in results]
    return list(zip(left_hand_side_of_rule, right_hand_side_of_rule, supports, confidences, lifts))

resultsinDataFrame = pd.DataFrame(inspect(results), columns = ['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
```
